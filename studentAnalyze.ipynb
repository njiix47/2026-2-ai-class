```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv('student.csv')
df.head()
```

```python
plt.hist(df['StudyHours'])
plt.xlabel('Study Hours')
plt.ylabel('Frequency')
plt.title('Study Hours Distribution')

plt.show()
```

```python
plt.scatter(df['StudyHours'], df['AssignmentScore'])
plt.show()
```

```python
result_mean = df.groupby('Result')['StudyHours'].mean()
```

```python
plt.bar(result_mean.index, result_mean.values)
plt.xticks([0, 1])

plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
plt.show()
```

```python
plt.scatter(df['Attendance'], df['AssignmentScore'])
plt.xlabel('AssignmentScore')
plt.ylabel('Attendancd and Assignment Score')
plt.show()
```

```python
plt.scatter(df['SleepHours'], df['Result'])
plt.show()
```

```python
result = df.groupby('Result')['Attendance'].mean()
```

```python
plt.bar(result.index, result.values)
plt.xticks([0, 1])

plt.ylabel('Average Attendance')
plt.title('Average Attendance by Result')
plt.show()
```
