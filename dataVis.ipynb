```python
import matplotlib.pyplot as plt

x_data = [2, 3, 4, 5]
y_data = [1, 3, 5, 7]

plt.ylim(0,8)
plt.plot(x_data, y_data, label = "data")

plt.title('Graph Title')
plt.xlabel('x_name')
plt.ylabel('y_name')
plt.legend()

plt.show()
```

```python
## 초미세먼지 데이터 시각화

import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel('dust_pm25.xlsx', index_col='area')
plt.figure(figsize=(9,3))

for year in range(2019, 2022) :
    yData = data[year]
    plt.plot(yData, marker = 's', label = year)

plt.xlabel('area')
plt.ylabel('micrometer')
plt.xticks(rotation = 45) # 기울기
plt.legend() # 범례 표기
plt.show()
```

```python
data2017 = data[2017]
plt.hist(data2017, bins = 10, label = 'pm2.5')

plt.xlabel('pm2.5')
plt.ylabel('frequncy')
plt.title('2017 Dust Histogram')
plt.legend()
plt.grid()
plt.show()
```
