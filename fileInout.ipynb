```python
# 엑셀파일 입출력

import pandas as pd

df = pd.read_excel('my_grade.xlsx')

add_df = pd.DataFrame([{
    '이름' : '자룡',
    '1차지필' : 63,
    '2차지필' : 75,
    '수행1' : 84,
    '수행2' : 79,
    '등급' : 'C' 
}])

modi_file = pd.concat([df, add_df], ignore_index=True)
modi_file.to_excel('new_grade.xlsx', index=False)

modi_file
```
