2023 날씨데이터 머신러닝

```python
import pandas as pd
weather = pd.read_excel('weather2023.xlsx', index_col='date')
weather.head()
```

```python
print(weather.isnull().sum())
```

```python
weather.drop(['location', 'loc_name', 'rain'], axis=1, inplace=True)
```

```python
weather.info()
```

describe() : 데이터프레임이나 특정열의 기초 통계값을 한 번에 확인하는 메서드, 각 열의 개수, 평균, 표준편차, 최솟값, 사분위소, 최댓값을 보여줌

```python
weather.describe()
```

corr() : 열과 열 사이의 상관계수를 계산하는 메서드

```python
weather.corr()
```

일조량과 전운량은 강한 음의 상관관계를 가진다. 구름의 양이 많을수록 일조량은 대체로 감소한다.

```python
import matplotlib.pyplot as plt
weather.hist(bins = 30) # bins = 30 : 데이터의 범위를 30개 구간으로 나눔
plt.show()
```

```python
x = weather['cloud']
y = weather['sunshine']
plt.scatter(x, y, color = 'green', alpha = 0.5)
plt.title('cloud-sunshime scatter')
plt.xlabel('cloud')
plt.ylabel('sunshine')
plt.show()
```

```python
x = weather['cloud']
y = weather['humi']
plt.scatter(x, y, color = 'blue', alpha = 0.5)
plt.title('cloud-humi scatter')
plt.xlabel('cloud')
plt.ylabel('humi')
plt.show()
```
